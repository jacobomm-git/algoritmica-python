### 4.6 Majúscules i minúscules

```python
def rle(text):
    """
    Aquesta funció retorna un text codificat segons run length encoding.
    
    Parameters
    ----------
    text: string
        text a codificar
        
    Returns
    -------
    text: string
        text codificat
    """
    t = ''
    lista = []
    for i in text:
      if(i in lista or len(lista)==0):
        lista.append(i)
      else:
        t = t + lista[0] + str(len(lista))
        lista.clear()
        lista.append(i)
    t = t + lista[0] + str(len(lista))
    
    return t
```

```python
assert rle("ABBBBNNNEEEDDDZZAAAAA") == 'A1B4N3E3D3Z2A5'
assert rle("BBBBBBBBBBBBBBBBBWWWWWWZAAA") == 'B17W6Z1A3'
```

       
La complexitat de l'algorisme és O(n).
